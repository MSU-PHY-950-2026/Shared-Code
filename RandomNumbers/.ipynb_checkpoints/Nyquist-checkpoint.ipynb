```python
import numpy
import matplotlib.pyplot as plt
```

```python
def samples(f, t, oversample=64):
  v = f(t)[:-1] # Evaluate function at each point except the last
  v = numpy.column_stack([v]*oversample).flatten() # Repeat each function value N times so it is flat for a while after each sample ("zero-order hold"). This is why we drop the last point above, since nothing is after it.
  t = numpy.concatenate([numpy.linspace(t[i], t[i+1], oversample) for i in range(0, len(t)-1)]) # Define times for each of the finer sampled points by interpolating
  return t,v
```

```python
plt.plot(*samples(numpy.sin, numpy.linspace(-10, 10, 50)))
```

```python
t0 = numpy.linspace(-10, 10, 50)
t, v = samples(numpy.sin, t0)
plt.plot(t,v)
plt.plot(t, numpy.sin(t))
```

```python
t0 = numpy.linspace(-10, 10, 25)
f0 = 1/(2*numpy.pi)
print('Signal frequency:', f0)
fnyquist = 1/(2*(t0[1] - t0[0]))
print('Nyquist frequency:', fnyquist)
falias1 = 2*fnyquist - f0
print('First alias frequency:', falias1)
falias2 = 2*fnyquist + f0
print('Second alias frequency:', falias2)
t, v = samples(numpy.sin, t0)
plt.plot(t,v)
plt.plot(t, numpy.sin(t))
#plt.plot(t, numpy.sin(-t/f0*falias1))
plt.plot(t, numpy.sin(t/f0*falias2))
plt.xlim(0, 4)
```

```python
t0 = numpy.linspace(-10, 10, 25)
t, y = samples(numpy.sin, t0)
print('Sample frequency:', 1/(t0[1] - t0[0]))
print('Nyquist frequency:', 1/(2*(t0[1] - t0[0])))
plt.plot(t,y)
plt.figure()
fft = numpy.fft.rfft(y)
plt.plot(numpy.fft.rfftfreq(len(t), t[1]-t[0]), numpy.log((fft * fft.conj()).real))
plt.xlim(0,4)
```

```python
t0 = numpy.linspace(-10, 10, 50)
t, y = samples(lambda x: numpy.sin(x*4), t0)
print('Sample frequency:', 1/(t0[1] - t0[0]))
print('Nyquist frequency:', 1/(2*(t0[1] - t0[0])))
print('Signal frequency:', 4*0.5/numpy.pi)
print('First image frequency:', 1/((t0[1] - t0[0])) - 4*0.5/numpy.pi)
print('Second image frequency:', 1/((t0[1] - t0[0])) + 4*0.5/numpy.pi)

plt.plot(t,y)
plt.figure()
fft = numpy.fft.rfft(y)
plt.plot(numpy.fft.rfftfreq(len(t), t[1]-t[0]), numpy.log((fft * fft.conj()).real))
plt.xlim(0,4)
```
