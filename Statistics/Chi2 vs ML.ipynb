```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def makeData(N,xmin,xmax,offset,slope):
    X = np.linspace(xmin,xmax,N)
    Y = offset+slope*X
    for i in range(0,N):
        Y[i] += np.random.randn()*np.sqrt(abs(Y[i]))
    return Y


# Make some data!!
# Gonna be a set of noisy data drawn from a linear function
xmin = 1 
xmax = 10
slope = 3

N = 50 # number of points
Xlo = np.linspace(xmin,xmax,N)
xline = [xmin,xmax]
yline = [offset+xmin*slope,offset+slope*xmax]
y_truelo = slope*Xlo
y_datalo = makeData(N,xmin,xmax,slope)

N = 500 # number of points
Xhi = np.linspace(xmin,xmax,N)
xline = [xmin,xmax]
yline = [xmin*slope,slope*xmax]
y_truehi = slope*Xhi
y_datahi = makeData(N,xmin,xmax,slope)

fig, axs = plt.subplots(nrows=1,ncols=2,figsize=(12,8))
axs[0].scatter(Xlo, y_datalo, s=40, label = "Random Data 50")
#axs[0].xlim(xmin-1,xmax+1)
axs[0].plot(xline,yline,linestyle="--",color="r", label="True Function")
axs[0].legend()
axs[1].scatter(Xhi, y_datahi, s=40, label = "Random Data 500")
#axs[1].xlim(xmin-1,xmax+1)
axs[1].plot(xline,yline,linestyle="--",color="r", label="True Function")
axs[1].legend()
plt.show()

```
